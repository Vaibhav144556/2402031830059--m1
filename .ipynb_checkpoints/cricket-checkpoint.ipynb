```python
import numpy as np

scores = np.array([
    [45, 67, 89, 34, 56],
    [23, 45, 67, 89, 90],
    [78, 56, 90, 67, 88],
    [34, 45, 23, 56, 78],
    [90, 87, 92, 88, 95],
    [56, 45, 67, 78, 89],
    [12, 23, 34, 45, 56],
    [67, 78, 89, 90, 91],
    [45, 56, 67, 78, 89],
    [88, 77, 66, 55, 44]
])
print(scores)
```

```python
# 1. Find total runs scored by each player
player1=sum(scores[0,:])
print("player 1 :",player1)
player2=np.sum(scores[1,:])
print("player 2 :",player2)
all_player_total=np.sum(scores,axis=1)
print("all players total : ",all_player_total)
```

```python
# average_runs of each player
avg_run=np.mean(scores,axis=1)
print("Average runs of each player : ",avg_run)
```

```python
# max and min score of all players
high_score=np.max(scores,axis=1)
print("maximum scores of players:",high_score)

law_score=np.min(scores,axis=1)
print("minimum scores of players",law_score)
```

```python
# in all of the matches print one-one player's maximum and minimum runs only
highest_score = np.max(scores)
print("Highest Score:", highest_score)

lowest_score = np.min(scores)
print("Lowest Score:", lowest_score)
```

```python
# print each player runs where avg>70
av_run=np.mean(scores,axis=1)
print(av_run)
runs=np.where(av_run>70)
print(runs)
```

```python
# total score of each match
match_total = np.sum(scores, axis=0)

print(match_total)
```
