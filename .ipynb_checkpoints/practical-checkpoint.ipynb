```python
vector = np.array([2,4,8,7,9,5,6,3])
print(np.sort(vector))
```

```python
matrix = np.array([[2,7,8,6],[1,7,8,6],[2,8,7,9]])
print(matrix)
```

```python
print(np.sort(matrix,axis=0))
```

```python
# descending
sorted_array = np.sort(matrix)
sort_array = np.sort(sorted_array,descending=True)
print(sort_array)
```
