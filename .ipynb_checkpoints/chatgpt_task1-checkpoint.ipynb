```python
import numpy as np

arr= np.array([[1, 2, 3],
                [4, 5, 6]])
print(arr)
```

```python
print(arr.shape)
```

```python
print("Rows:", arr.shape[0])
print("Columns:", arr.shape[1])
```

```python
print(arr[1, 1])
print(arr[0])
print(arr[:, 1])
```

```python
print(type(arr))
```

```python
print(arr.dtype)
```
